# FinAI Edge - Quantitative Research Sandbox
Use this notebook to experiment with Modern Portfolio Theory, covariance matrices, and risk analytics before migrating production logic to `portfolio_engine.py`.

```python
import matplotlib.pyplot as plt
import numpy as np
import yfinance as yf

# Fetch sample data
tickers = ["RELIANCE.NS", "TCS.NS", "INFY.NS", "HDFCBANK.NS"]
data = yf.download(tickers, period="2y", auto_adjust=True)["Close"]
returns = np.log(data / data.shift(1)).dropna()

returns.head()
```

```python
# Covariance and Correlation Analysis
cov_matrix = returns.cov() * 252
corr_matrix = returns.corr()

import seaborn as sns

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Asset Correlation Heatmap")
plt.show()
```
